A base contém medidas de flores de três espécies: *setosa*, *versicolor* e *virginica*. Cada registro representa uma flor; as variáveis incluem comprimento e largura das sépalas e das pétalas.

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df_iris = iris.frame
df_iris["species"] = df_iris["target"].map(dict(enumerate(iris.target_names)))
df_iris = df_iris.drop(columns="target")

df_iris.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df_iris, x="species")
plt.title("Flores por espécie")
plt.xlabel("Espécie")
plt.ylabel("Quantidade de flores")
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=df_iris, x='sepal length (cm)', y='petal length (cm)', hue='species')
plt.title('Comparação do comprimento de sépalas e pétalas')
plt.xlabel('Comprimento da sépala')
plt.ylabel('Comprimento da pétala')
plt.tight_layout()
plt.show()
```

```python
O resultado me mostrou que, com exceção da setosa, quando a sépala aumentava de tamanho, a pétala tendia a aumentar também. Para verificar se era verdade ou não, fiz um corr:
```

```python
numerico = df_iris.groupby('species')[['sepal length (cm)', 'petal length (cm)']]
correlacao = numerico.corr()
sns.heatmap(correlacao, annot=True, cmap='vlag', center=0, fmt=".2f")
plt.title('Comparação do comprimento de sépalas e pétalas')
plt.tight_layout()
plt.show()
```

```python

E ele me mostrou que a correlação entre comprimento da sépala e comprimento da pétala da setosa é de 0.27, ou seja, não muito alto. Porém, no caso da versicolor é de 0.75, e da virginica é de 0.86. Mostrando uma grande correlação entre o aumento do comprimento da sépala e da pétala. Embora não necessariamente exista uma causalidade entre elas.

## Entrega

Organize as respostas no próprio notebook, abaixo de cada pergunta, incluindo:
- código usado na análise;
- tabelas ou estatísticas relevantes;
- gráficos com título e nomes dos eixos;
- uma interpretação escrita para cada pergunta.

**Conclusão final:** escreva um pequeno parágrafo resumindo o que descobriu sobre a base Iris e cite uma limitação da sua análise.

**Critério principal:** valorize a qualidade da investigação e da interpretação, não a quantidade de gráficos. Explique como cada resultado ajuda a responder à pergunta.
```
